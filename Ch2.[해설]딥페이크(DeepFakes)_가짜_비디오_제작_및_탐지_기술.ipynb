## <b>주제: 딥페이크(DeepFakes) 가짜 비디오 제작 및 탐지 기술</b>
------------------------

### <b>주제 설명</b>

* <b>딥페이크(Deepfake)</b>는 특정한 인물이 등장하는 이미지 혹은 영상을 생성하는 기술입니다. 딥페이크 기술은 악용되었을 때 많은 사회적 이슈를 발생시킬 수 있으나, 영화 스턴트 촬영 및 CG 합성 등에서 긍정적으로 활용할 부분이 많은 기술입니다.

* 본 실습에서는 다음의 3가지 딥페이크(Deepfake) 핵심 기술을 학습합니다.
    * <b>얼굴 표정 재연(Face Reenactment)</b>: 원하는 인물이 특정한 행동을 흉내 내도록 만듭니다.
    * <b>얼굴 교체(Face Swapping)</b>: 동영상의 얼굴 부분에 원하는 인물의 정체성(identity)을 삽입합니다.
    * <b>딥페이크 탐지(Deepfake Detection)</b>: 특정한 동영상이 Deepfake인지의 여부를 탐지합니다.

#### <b>문제 4. 이미지와 비디오 전처리하기</b>

````
import imageio
from skimage.transform import resize

# 이미지 불러오기
source_image = imageio.imread('source.png')

# 동영상 불러오기
reader = imageio.get_reader('reference.mp4') # 동영상 리더(reader) 객체 초기화
fps = reader.get_meta_data()['fps']
frames = [] # 동영상의 모든 프레임을 담을 객체
for im in reader:
    frames.append(im)
reader.close()

# 이미지와 동영상을 (256 X 256) 해상도로 변경
source_image = resize(source_image, (256, 256))[..., :3] # 알파(alpha) 값이 있다면 제거
frames = [resize(frame, (256, 256)) for frame in frames]

# 해상도 확인하기
print('이미지 해상도:', source_image.shape)
print('동영상 해상도:', frames[0].shape) # 첫 번째 프레임의 크기 출력
````

#### <b>문제 5. 이미지와 동영상을 화면에 출력하기</b>

````
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML


# 이미지와 동영상을 함께 출력하는 메서드
def display(source_image, frames, predictions=None):
    fig = plt.figure(figsize=(8 + 4 * (predictions is not None), 4)) # 하나의 그림(figure) 생성
    images = []
    for i in range(len(frames)):
        cols = [source_image] # 왼쪽에는 소스 이미지(image)
        cols.append(frames[i]) # 오른쪽에는 현재의 비디오 프레임(frame)
        if predictions is not None: # 예측 결과(prediction)가 있는 경우에는 함께 출력
            cols.append(predictions[i])
        image = plt.imshow(np.concatenate(cols, axis=1), animated=True) # 가로 너비(width)를 기준으로 합치기
        images.append([image])

    plt.close()
    ani = animation.ArtistAnimation(fig, images, interval=50, repeat_delay=1000) # 애니메이션(animation) 객체 생성
    return ani

HTML(display(source_image, frames).to_html5_video()) # HTML 비디오(video) 형식으로 화면에 출력
````

## <b>Step 2. Single Image 얼굴 교체(Face Swapping)</b>

* <b>Step 2</b>로 넘어가기 전에 <b>[런타임]</b> - <b>[런타임 다시 시작]</b> 버튼을 눌러 런타임을 초기화합니다.

#### <b>핵심 아이디어 설명</b>

* target video에서 물체(얼굴)를 구성하는 부분(part)들을 분할(segmentation)합니다.
* target video의 <b>모션(motion)은 유지한 상태로</b> source image에서 특정 부분(part)의 정체성(identity)을 가져와 적용할 수 있습니다.
    * 결과적으로 <b>한 장의 이미지만 가지고도 가짜 비디오(fake video)를 제작</b>할 수 있습니다.

image.png

* 본 모델은 영상 내 물체(object)를 구성하는 부분(part)들을 <b>자동으로 분할(segmentation)</b>할 수 있습니다.
    * <b>예시)</b> 어디가 팔 부분인지, 다리 부분인지 학습 데이터에 명시하지 않아도 됩니다.
    * <b>예시)</b> 어디가 눈동자 부분인지, 입술 부분인지 학습 데이터에 명시하지 않아도 됩니다.
* 학습된 이후에는 모션(motion)은 유지한 상태로 특정한 부분(part)이 선택적으로 바뀌도록 할 수 있습니다.
    * <b>예시)</b> 얼굴에서 눈동자만 다른 사람의 눈동자로 변경할 수 있습니다.

image.png

* 학습 시기에 <b>동일한 비디오</b>의 두 개의 프레임(frame)을 선택합니다.
    * target frame의 segments를 예측한 뒤에, source frame을 활용해 target frame을 reconstruct합니다.

````
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML


# 이미지와 동영상을 함께 출력하는 메서드
def display(source_image, frames, predictions=None):
    fig = plt.figure(figsize=(8 + 4 * (predictions is not None), 4)) # 하나의 그림(figure) 생성
    images = []
    for i in range(len(frames)):
        cols = [source_image] # 왼쪽에는 소스 이미지(image)
        cols.append(frames[i]) # 오른쪽에는 현재의 비디오 프레임(frame)
        if predictions is not None: # 예측 결과(prediction)가 있는 경우에는 함께 출력
            cols.append(predictions[i])
        image = plt.imshow(np.concatenate(cols, axis=1), animated=True) # 가로 너비(width)를 기준으로 합치기
        images.append([image])

    plt.close()
    ani = animation.ArtistAnimation(fig, images, interval=50, repeat_delay=1000) # 애니메이션(animation) 객체 생성
    return ani

HTML(display(source_image, frames).to_html5_video()) # HTML 비디오(video) 형식으로 화면에 출력
````

#### <b>문제 15. 얼굴 분할(Face Segmentation) 결과 확인하기</b>

````
import torch
import torch.nn.functional as F

import numpy as np
import matplotlib.patches as mpatches

import warnings
warnings.filterwarnings("ignore") # 오래 된(deprecated) 코드 경고 메시지 보이지 않도록 하기


# 파싱 네트워크(parsing network)를 이용해 얼굴을 여러 부분으로 분할(segmentation)하여 시각화
def visualize_segmentation(image, network, supervised=False, hard=True, colormap='gist_rainbow'):
    # 네트워크를 이용해 마스크 예측(prediction)
    with torch.no_grad():
        # 이미지의 형태를 바꾸어 PyTorch의 Tensor 형태로 변경: [batch_size, channel, height, width]
        input = torch.tensor(image[np.newaxis].astype(np.float32)).permute(0, 3, 1, 2).cuda()
        input = F.interpolate(input, size=(512, 512))
        input = (input - network.mean) / network.std
        mask = torch.softmax(network(input)[0], dim=1) # 네트워크에 입력하여 마스크(mask) 확인
        mask = F.interpolate(mask, size=image.shape[:2]) # 마스크와 이미지의 크기를 동일하도록

    if hard:
        mask = (torch.max(mask, dim=1, keepdim=True)[0] == mask).float()

    # 색상 마스크(color mask) 만들기
    colormap = plt.get_cmap(colormap)
    num_segments = mask.shape[1]
    mask = mask.squeeze(0).permute(1, 2, 0).cpu().numpy() # 이미지의 형태 변경: [height, width, channel]
    color_mask = 0
    patches = []

    # 마스크 차원 출력
    print('마스크 차원:', mask.shape)

    # 세그먼트(segment)의 개수만큼 반복하며 각각 다른 색상 적용
    for i in range(num_segments):
        if i != 0:
            color = np.array(colormap((i - 1) / (num_segments - 1)))[:3]
        else:
            color = np.array((0, 0, 0))
        patches.append(mpatches.Patch(color=color, label=str(i)))
        color_mask += mask[..., i:(i + 1)] * color.reshape(1, 1, 3)
    
    # 두 장의 이미지가 담길 수 있는 그림(figure) 생성
    fig, ax = plt.subplots(1, 2, figsize=(12, 6))

    ax[0].imshow(color_mask)
    ax[1].imshow(0.3 * image + 0.7 * color_mask)
    ax[1].legend(handles=patches)
    ax[0].axis('off')
    ax[1].axis('off')
    plt.show()

visualize_segmentation(source_image, face_parser, supervised=True, hard=True, colormap='tab20')
````

#### <b>문제 22. 바운딩 박스(Bounding Box) 계산 메서드</b>

* dlib를 통해 찾은 얼굴을 <b>조금 더 큰 크기의 바운딩 박스로 변환</b>합니다.
* <b>파라미터(parameters)</b>
    * face: 이미지내 얼굴이 존재하는 위치(position)
    * width: 원본 이미지의 너비
    * height: 원본 이미지의 높이
    * scale: 원본 얼굴보다 얼마나 큰 바운딩 박스를 찾을지
* <b>반환 값(return value)</b>
    * x, y, bounding_box_size: 왼쪽 위(top-left) 좌표와 크기

````
import dlib
import cv2
import matplotlib.pyplot as plt


def get_boundingbox(face, width, height, scale=1.3):
    x1 = face.left()
    y1 = face.top()
    x2 = face.right()
    y2 = face.bottom()
    size_bb = int(max(x2 - x1, y2 - y1) * scale) # 바운딩 박스의 크기 설정
    center_x, center_y = (x1 + x2) // 2, (y1 + y2) // 2 # 중간 지점

    # 왼쪽 위(top-left) 좌표 설정
    x1 = max(int(center_x - size_bb // 2), 0)
    y1 = max(int(center_y - size_bb // 2), 0)
    # 바운딩 박스의 크기가 너무 크다면
    size_bb = min(width - x1, size_bb)
    size_bb = min(height - y1, size_bb)

    return x1, y1, size_bb
````

````
from PIL import Image
from dataset.transform import xception_default_data_transforms


def preprocess_image(image):
    # BGR 형식을 RGB 형식으로 변환
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # 테스트(test) 시기에 사용하는 PyTorch transforms 객체 불러오기
    preprocess = xception_default_data_transforms['test']

    # PIL image 객체로 바꾼 뒤에 PyTorch 객체로 변환
    preprocessed_image = preprocess(Image.fromarray(image))

    # PyTorch 형식에 맞게 앞 부분에 배치 추가 [channel, height, with ] → [1, channel, height, width]
    preprocessed_image = preprocessed_image.unsqueeze(0).cuda()

    return preprocessed_image
````

#### <b>문제 24. 딥페이크 판별 모델 불러오기</b>

* Xception 네트워크 아키텍처로 학습된 모델을 준비합니다.

````
import torch.nn as nn
import torch
from network.models import model_selection
import warnings
warnings.filterwarnings("ignore") # 소스코드 변경 경고(SourceChangeWarning) 메시지 보이지 않도록 하기

model_path = 'faceforensics++_models_subset/full/xception/full_raw.p'

# 모델 불러오기(model loading)
model, *_ = model_selection(modelname='xception', num_out_classes=2)
model = torch.load(model_path)
model = model.cuda()
````

#### <b>문제 25. 네트워크에 넣어 결과를 확인하는 메서드</b>

* 이미지 Numpy 객체를 받아 전처리 이후에 네트워크에 넣어 판별 결과를 확인합니다.
* <b>파라미터(parameters)</b>
    * image: PyTorch의 Tensor 객체
    * model: 탐지(detection) 목적의 네트워크 모델
* <b>반환 값(return value)</b>
    * prediction: 탐지 결과 (Fake = 1, Real = 0)

````
def predict_with_model(image, model, post_function=nn.Softmax(dim=1)):
    # 전처리(preprocess) 수행
    preprocessed_image = preprocess_image(image)

    # 모델 예측(model prediction)
    output = model(preprocessed_image)
    output = post_function(output)

    # 판별 결과(prediction)와 확률 값을 반환
    _, prediction = torch.max(output, 1)
    prediction = float(prediction.cpu().numpy())

    return int(prediction), output
````

````
image = cv2.imread("./source.png")

face_detector = dlib.get_frontal_face_detector()
face = face_detector(image) # 얼굴 탐지 수행
face = face[0] # 이미지 내에서 첫 번째 얼굴 선택

height, width = image.shape[:2]
x, y, size = get_boundingbox(face, width, height) # 바운딩 박스(bounding box) 찾기
cropped = image[y:y + size, x:x + size]

prediction, output = predict_with_model(cropped, model)
label = 'Fake' if prediction == 1 else 'Real'
output_list = ['{0:.2f}'.format(float(x)) for x in output.detach().cpu().numpy()[0]]

print("판별 결과:", label)
print("확률 값 출력:", output_list)
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.show()

# 출력용 변수 초기화
x = face.left()
y = face.top()
w = face.right() - x
h = face.bottom() - y

font_face = cv2.FONT_HERSHEY_SIMPLEX
thickness = 2
font_scale = 0.5
color = (0, 255, 0) if prediction == 0 else (0, 0, 255)

print("[ 이미지에 판별 결과 그리기 ]")
# 판별 결과를 텍스트 형태로 그리기
cv2.putText(image, str(output_list) + '=>' + label, (x, y + h + 20), font_face, font_scale, color, thickness, 2)

# 얼굴 주위로 바운딩 박스(bounding box) 그리기
cv2.rectangle(image, (x, y), (x + w, y + h), color, 2)

# 바운딩 박스가 포함된 이미지 그리기
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.show()
````

#### <b>문제 26. 비디오에서 얼굴을 검출하여 결과 확인하기</b>

* <b>파라미터(parameters)</b>:
    * video_path: 비디오 파일(video file)의 경로
    * model_path: 모델 파일(model file)의 경로
    * output_path: 출력 비디오(output video)가 저장될 경로

````
import os
from os.path import join

video_path = "face_swapping.mp4"
model_path = 'faceforensics++_models_subset/full/xception/full_raw.p'
output_path = 'detection_output'

print('Starting: {}'.format(video_path))

# 원본 영상을 읽기 위한 Reader 객체
reader = cv2.VideoCapture(video_path)

# 판별 결과를 쓰기 위한 Writer 객체
video_fn = video_path.split('/')[-1].split('.')[0] + '.avi' # 저장될 결과 영상 파일 이름
os.makedirs(output_path, exist_ok = True)
fourcc = cv2.VideoWriter_fourcc(*'MJPG')
fps = reader.get(cv2.CAP_PROP_FPS) # 원본 영상의 FPS를 사용
num_frames = int(reader.get(cv2.CAP_PROP_FRAME_COUNT)) # 원본 영상의 전체 프레임 개수
writer = None

# dlib 얼굴 탐지 라이브러리
face_detector = dlib.get_frontal_face_detector()

# 텍스트 출력 관련 변수
font_face = cv2.FONT_HERSHEY_SIMPLEX
thickness = 2
font_scale = 0.5

# 처리한 프레임 수
current_frame = 0

# 동영상 리더(reader) 객체가 열려있다면 반복
while reader.isOpened():
    # 동영상에서 현재의 프레임(frame) 화면 캡처
    _, image = reader.read()
    if image is None:
        break
    current_frame += 1

    # 탐지 경과에 따라 출력
    if current_frame % 20 == 0:
        print(f"[Detection] Processed Frames: [{current_frame}/{num_frames}]")

    # 이미지 크기 받아오기
    height, width = image.shape[:2]

    # Output Writer 초기화(initialization)
    if writer is None:
        writer = cv2.VideoWriter(join(output_path, video_fn), fourcc, fps, (height, width)[::-1])

    # dlib를 이용해 얼굴 탐지
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    faces = face_detector(gray, 1)

    # 최소한 1개 이상의 얼굴이 탐지 된 경우
    if len(faces):
        # 가장 큰 얼굴(the biggest face)만 사용하기
        face = faces[0]

        """ 먼저 예측(prediction)을 수행합니다. """
        # dlib을 이용해 얼굴 부분을 잘라낸 뒤에 바운딩 박스의 크기를 키웁니다.
        x, y, size = get_boundingbox(face, width, height)
        cropped_face = image[y:y + size, x:x + size]

        # 실제로 네트워크를 이용해 예측을 진행합니다.
        prediction, output = predict_with_model(cropped_face, model)
        label = 'Fake' if prediction == 1 else 'Real'
        output_list = ['{0:.2f}'.format(float(x)) for x in output.detach().cpu().numpy()[0]]

        """ 예측 결과를 화면에 출력합니다. """
        # 출력용 변수 초기화
        x = face.left()
        y = face.top()
        w = face.right() - x
        h = face.bottom() - y

        color = (0, 255, 0) if prediction == 0 else (0, 0, 255)

        # 판별 결과를 텍스트 형태로 그리기
        cv2.putText(image, str(output_list) + '=>' + label, (x, y + h + 20), font_face, font_scale, color, thickness, 2)

        # 얼굴 주위로 바운딩 박스(bounding box) 그리기
        cv2.rectangle(image, (x, y), (x + w, y + h), color, 2)

    # 처리된 이미지를 결과 영상에 쓰기
    writer.write(image)

if writer is not None:
    writer.release()
    print('Finished!')
else:
    print('The input video file was empty')
````

#### <b>핵심 아이디어 설명</b>

    * 성공적인 결과물 생성을 위하여 학습 이미지는 일반적으로 최소 <b>1,000장 이상 필요</b>합니다.

* <b>학습 단계(Training Phase)</b>
    * 인코더(encoder)는 공유하고 <b>디코더(decoder)는 얼굴 정체성(identity)마다 다르게 학습</b>합니다.

    * 두 개의 디코더는 두 개의 비디오 각각에서 얼굴의 정체성(identity)을 학습하게 됩니다.

    * 변환을 수행할 때는 인코더(encoder) A에 넣어 얻은 결과를 디코더(decoder) B에 넣습니다.
    * 결과적으로 얼굴의 정체성(identity)만 인물 B의 정체성으로 교체됩니다.

* 학습 도중에 진행 상황을 시각화하기 위해 클래스를 정의합니다.

````
from time import sleep, time
from IPython.display import display, HTML
from threading import Thread
import base64

image_path = "faceswap/training_preview.jpg"
!touch '{image_path}'


class PreviewImg:
    def __init__(self):
        self.cancel = False
        self.fig = display(HTML('<img src="https://dummyimage.com/1024x618/000/ffffff&text=Preview+Loading..." />'), display_id=True)
        print("디스플레이를 생성합니다.")

    def update(self):
        print("디스플레이를 업데이트합니다.")
        # 디스플레이 업데이트 코드 작성
        with open(image_path, 'rb') as image:
            encoded = str(base64.b64encode(image.read()))[2:-1]
            self.fig.update(HTML('<img src="data:image/jpg;base64,' + encoded + '" width="1024" height="618" />'))

    # 쓰레드(thread)에 의해서 실행될 메서드
    def task(self):
        while not self.cancel:
            sleep(5 * 60) # 주기적으로 업데이트
            if not self.fig:
                print("이미지가 존재하지 않습니다.")
                break
            self.update()
````

* 모델 학습을 진행합니다.
* 백엔드(backend) GPU로 <b>NVIDIA를 선택</b>합니다.
* GPU 자원이 부족해 오류가 발생하는 경우 배치 사이즈(batch size) 파라미터를 줄입니다.
* 일반적으로 그럴싸한 결과를 얻기 위해서 0.01 가량의 손실(loss) 값에 도달해야 합니다.
* 이를 위해 Google Colab 기준으로 <b>15시간 이상의 반복적인 학습</b>이 필요합니다.
    * 중간에 GPU 사용량 초과로 Colab이 종료되는 경우, Google Drive에 저장된 모델을 토대로 학습을 재개하면 됩니다.
* <b>(참고)</b> 새로운 데이터셋으로 학습하는 경우, 새로운 폴더를 지정해서 모델이 저장될 수 있도록 합니다.

````
# 기본적으로 모델 파일이 이미 존재하면 해당 모델을 이용해 이어서 학습을 진행하므로, 새로운 타겟 이미지로 학습하는 경우 새로운 모델 파일 경로를 정의하기
!echo "프리뷰(Preview) 이미지의 크기는 실제보다 작으므로, 우클릭 이후에 [새 탭에서 이미지 열기]로 확인하세요."

prev_img = PreviewImg()
thread = Thread(target=prev_img.task)
thread.start()

# 모델 학습 진행
!python3 faceswap/faceswap.py train -A 'source_images' -B 'target_images' -m '/content/drive/My Drive/Colab Notebooks/faceswap_models' -t 'villain' -bs 8 -s 330 -w -nl
````
