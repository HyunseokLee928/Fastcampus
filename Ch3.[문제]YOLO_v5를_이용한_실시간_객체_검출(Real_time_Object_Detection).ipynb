## <b>Step 2. YOLO v5 아키텍처 핵심 정리</b>

* YOLO v5 아키텍처의 핵심 내용을 공부합니다.
* YOLO v1부터 YOLO v4까지의 <b>발전 과정</b>을 차례대로 확인합니다.

#### <b>(이론) YOLO v1 설명</b>

* <b>참고</b>: [You Only Look Once: Unified, Real-Time Object Detection (CVPR 2016)](https://arxiv.org/abs/1506.02640)
* <b>45 FPS</b>의 성능을 보여 실시간 객체 탐지(real-time object detection)에 적합합니다.
* YOLO v1는 이미지를 받아 특징(feature)을 추출(extraction)한 뒤에 <b>7 X 7 그리드(grid) 셀마다 클래스를 예측</b>합니다.
* 각 그리드 셀에서 2개의 바운딩 박스(bounding box) 정보와 confidence score 정보를 예측합니다.


image.png

* YOLO는 객체 탐지(object detection)에 필요한 많은 구성요소를 <b>하나의 네트워크로 통합</b>했습니다.
    * R-CNN 계열의 논문에 비하면 전체 아키텍처가 비교적 간단합니다.
* 하나의 이미지에 대한 전반적인(global) 특징을 추출한 뒤에, 결과적으로 각 그리드에 대해서 예측 결과를 구합니다.
    * 출력 텐서의 크기: <b>7 X 7 X 30</b> = 7 X 7 X (2개(바운딩 박스 개수) X 5개(바운딩 박스 좌표와 confidence score) + 20개(클래스 개수)) 
* 이후에 <b>NMS (Non-Maximum Suppression)</b>을 이용해 하나의 객체(object)에 대하여 가장 confidence score가 높은 바운딩 박스만 남깁니다.

image.png

#### <b>(이론) YOLO v2 설명</b>

* <b>참고</b>: [YOLO9000: Better, Faster, Stronger (CVPR 2017)](https://arxiv.org/abs/1612.08242)
* YOLO v1처럼 속도는 여전히 빠릅니다.
* <b>Faster</b>: 속도 측면에서 높은 이점이 있는 아키텍처인 <b>Darknet-19</b>를 제안합니다.
* <b>Stronger</b>: dataset combination과 joint training을 이용해 9,000개의 클래스(class)를 분류할 수 있는 <b>YOLO 9000</b>을 제안합니다.
* <b>Better</b>: 구체적으로 다음과 같은 다양한 테크닉을 적용하여 정확도(mAP)를 많이 향상시켰습니다.

image.png

* <b>K-means 클러스터링</b>을 이용해 ground-truth box와 유사한 anchor box를 학습한 뒤에 사용합니다.
    * <b>특정한 데이터셋에 적합한 anchor box</b>를 찾을 수 있기 때문에 성능이 개선됩니다.
    * 더 적은 개수의 anchor box를 사용해도, 고정된 anchor box (Faster R-CNN에서의)보다 더 좋은 성능을 보입니다.

image.png

* YOLO v2에서는 YOLO v1과는 다르게 fully-connected layer를 사용하지 않습니다.
* 하나의 그리드(grid) 셀마다 5개의 anchor box를 이용해 예측을 수행합니다.
    * 하나의 그리드 안에서 <b>anchor box 별로 클래스 확률(probability)을 예측</b>합니다.

#### <b>(이론) YOLO v3 설명</b>

* <b>참고</b>: [YOLOv3: An Incremental Improvement (2018)](https://arxiv.org/abs/1804.02767)
* YOLO v2에 작은 변화들(changes)을 추가하여 성능을 조금 더 개선했습니다.
* <b>Better</b>: 조금 더 깊은 Darknet-53 아키텍처를 이용하여 정확도를 많이 개선합니다. (깊지만 ResNet보다는 빠르게 동작)
* YOLO v2의 Darknet-19과 YOLO v3의 Darknet-53을 비교한 것은 다음과 같습니다.
    * <b>Darknet-19</b>의 아키텍처는 VGG 모델과 유사하지만, 연산 수가 적다는 점에서 빠르게 동작합니다.
    * <b>Darknet-53</b>의 아키텍처는 shortcut connections를 적용하여 정확도가 많이 개선되었습니다.

image.png

* YOLO 9000과 유사하게 <b>바운딩 박스 예측(bounding box prediction)</b>을 진행합니다.
* 각 바운딩 박스마다 logistic regression을 이용해 <b>사물이 존재하는지(objectness score)</b>를 예측합니다.
    * 다른 bounding box prior보다 ground truth와 overlapping이 많이 되는 bounding box prior에 1의 값을 부여합니다.
* 또한 softmax를 사용하지 않고 여러 개의 독립적인 logistic classifier들을 사용합니다. (binary cross-entropy 사용)

image.png

#### <b>(이론) YOLO v4 설명</b>

* <b>참고</b>: [YOLOv4: Optimal Speed and Accuracy of Object Detection (2020)](https://arxiv.org/abs/2004.10934)
* YOLO v3까지의 저자(Joseph Redmon)가 작성한 논문이 아닙니다.
* 잘 동작한다고 알려진 다양한 기술들을 YOLO 아키텍처에 적용해보며 정확도를 더욱 향상시킨 논문입니다.
* <b>백본(Backbone)</b>: 정확도가 높으며 빠르게 동작하는 CSPDarknet53를 사용합니다.
    * 512 X 512의 고해상도 이미지를 입력으로 받습니다.
    * 에너지 소비량이 적으면서 성능이 우수한 CSPNet (CVPR 2020) 논문에서 제안된 아이디어를 적용합니다.
* <b>넥(Neck)</b>: SPP, PAN
* <b>헤드(Head)</b>: YOLO v3
* 이 밖에도 <b>다양한 테크닉</b>을 적용: ReLU, CutOut, MixUp, CutMix, DropOut, Batch Normalization, Residual Connection 등
* YOLO v3와 비교했을 때, 속도는 여전히 빠르면서 정확도(accuracy)를 크게 향상시켰습니다.

#### <b>(이론) 정확도(Precision)와 재현율(Recall)</b>

* <b>Precision</b> = 올바르게 탐지한 물체의 수 / 모델이 탐지한 물체의 수
* <b>Recall</b> = 올바르게 탐지한 물체의 수 / 실제 정답 물체의 수
* 일반적으로 정확도(Precision)와 재현율(Recall)은 반비례 관계를 가집니다.
    * 무작정 모든 영역에 물체가 존재한다고 판단하는 경우 재현율은 높지만, 정확도는 낮을 것입니다.
* 따라서 평균 정확도(average precision)로 성능을 평가하는 경우가 많습니다.
* <b>참고 자료</b>: https://www.catalyzex.com/paper/arxiv:1607.03476

image.png

* <b>IOU (Intersection Over Union)</b>는 두 바운딩 박스(bounding box)가 겹치는 비율을 의미합니다.
* <b>mAP\@0.5</b>는 정답과 예측의 IOU가 50% 이상일 때 정답으로 판정하겠다는 의미입니다.
* <b>mAP\@0.5:0.95</b>는 0.5에서 0.95까지 IOU를 0.05씩 증가시키며 계산한 값의 평균을 의미합니다.
